```python
import pandas as pd
import os

# Define the path to the local CSV file
data_file_path = '/content/flood.csv'

# Load the dataset into a pandas DataFrame
try:
    df = pd.read_csv(data_file_path)
    print("\nSuccessfully loaded the dataset into a DataFrame.")
    display(df.head())
except FileNotFoundError:
    print(f"Error: '{data_file_path}' not found. Please ensure the file is uploaded to /content/.")
except Exception as e:
    print(f"An error occurred while loading the dataset: {e}")
```

```python
# Separate features (X) and target variable (y)
X = df.drop('FloodProbability', axis=1)
y = df['FloodProbability']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
display(X.head())
display(y.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)

# Convert the scaled features back to a DataFrame for better readability (optional)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Scaled Features (X_scaled) shape:", X_scaled_df.shape)
display(X_scaled_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.decomposition import PCA

# Initialize PCA with 10 components
pca = PCA(n_components=10)

# Fit PCA on the scaled training data and transform both training and testing data
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print("Shape of X_train after PCA:", X_train_pca.shape)
print("Shape of X_test after PCA:", X_test_pca.shape)

# Optionally, you can examine the explained variance ratio
print("Explained variance ratio of top 10 components:", pca.explained_variance_ratio_)
print("Cumulative explained variance:", sum(pca.explained_variance_ratio_))
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("Successfully imported all necessary machine learning libraries.")

# Define models
models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(random_state=42),
    'Lasso': Lasso(random_state=42),
    'ElasticNet': ElasticNet(random_state=42),
    'K-Neighbors Regressor': KNeighborsRegressor(),
    'Decision Tree Regressor': DecisionTreeRegressor(random_state=42),
    'Gradient Boosting Regressor': GradientBoostingRegressor(random_state=42),
    'XGBoost Regressor': XGBRegressor(random_state=42),
    'LightGBM Regressor': LGBMRegressor(random_state=42),
    'CatBoost Regressor': CatBoostRegressor(random_state=42, verbose=0),
    'Support Vector Regressor': SVR()
}

print("Successfully defined the models dictionary.")
```

```python
model_names = []
r2_scores = []
mae_scores = []
mse_scores = []

for name, model in models.items():
    print(f"\nTraining and evaluating: {name}")

    # Train the model
    model.fit(X_train_pca, y_train)

    # Make predictions
    y_pred = model.predict(X_test_pca)

    # Calculate metrics
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    # Store results
    model_names.append(name)
    r2_scores.append(r2)
    mae_scores.append(mae)
    mse_scores.append(mse)

    print(f"R-squared: {r2:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"MSE: {mse:.4f}")
```

```python
results_df = pd.DataFrame({
    'Model': model_names,
    'R-squared': r2_scores,
    'MAE': mae_scores,
    'MSE': mse_scores
})

# Sort by R-squared in descending order to easily find the best model
results_df_sorted = results_df.sort_values(by='R-squared', ascending=False).reset_index(drop=True)

print("\nModel Performance Comparison (Sorted by R-squared):")
display(results_df_sorted)

# Identify the best performing model based on R-squared
best_model = results_df_sorted.iloc[0]
print(f"\nBest Performing Model (based on R-squared):\n{best_model}")
```

```python
import joblib
import os

# Create a directory to store the models if it doesn't exist
output_dir = 'trained_models'
os.makedirs(output_dir, exist_ok=True)

print(f"Saving models to directory: {output_dir}")
saved_model_paths = []

for name, model in models.items():
    # Generate a filename for the model
    filename = f"{name.replace(' ', '_').replace('-', '_').lower()}.pkl"
    filepath = os.path.join(output_dir, filename)

    # Save the model to a .pkl file
    joblib.dump(model, filepath)
    saved_model_paths.append(filepath)
    print(f"Saved {name} to {filepath}")

print("\nAll models have been saved.")
print("Paths to saved models:")
for path in saved_model_paths:
    print(path)
```
